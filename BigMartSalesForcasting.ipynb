```python
# ══════════════════════════════════════════════════════════════════════════════
#  NEW NOTEBOOK CELLS
#  Add these after Cell 100 (mean_absolute_error) in your notebook.
#  They slot in BEFORE Cell 101 "### Prediction on Unseen Data"
# ══════════════════════════════════════════════════════════════════════════════


# ─────────────────────────────────────────────────────────────────────────────
# SECTION A  ·  Better Evaluation  (R², RMSE, Actual vs Predicted scatter)
# Add after Cell 100: mean_absolute_error(y_test, y_pred)
# ─────────────────────────────────────────────────────────────────────────────

# Cell A1 — full metric suite
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker

mae  = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2   = r2_score(y_test, y_pred)

print("=" * 38)
print(f"  MAE  : ₹{mae:>10,.2f}")
print(f"  RMSE : ₹{rmse:>10,.2f}")
print(f"  R²   :  {r2:>10.4f}")
print("=" * 38)
print(f"\n  Model explains {r2*100:.1f}% of sales variance.")
print(f"  Average prediction error: ±₹{mae:,.0f}")


# Cell A2 — Actual vs Predicted scatter plot
fig, ax = plt.subplots(figsize=(7, 5))
fig.patch.set_facecolor('#0d0d0d')
ax.set_facecolor('#131313')
for sp in ax.spines.values():
    sp.set_visible(False)

# scatter
ax.scatter(y_test, y_pred, alpha=0.35, s=12,
           color='#ff6b35', linewidths=0, zorder=3)

# perfect-prediction line
lim_min = min(y_test.min(), y_pred.min()) * 0.95
lim_max = max(y_test.max(), y_pred.max()) * 1.05
ax.plot([lim_min, lim_max], [lim_min, lim_max],
        color='#4a7fcb', linewidth=1.5, linestyle='--',
        label='Perfect prediction', zorder=4)

# ±MAE band
ax.fill_between([lim_min, lim_max],
                [lim_min - mae, lim_max - mae],
                [lim_min + mae, lim_max + mae],
                alpha=0.08, color='#ff6b35',
                label=f'±MAE band  (₹{mae:,.0f})', zorder=2)

ax.set_xlim(lim_min, lim_max)
ax.set_ylim(lim_min, lim_max)
ax.set_xlabel('Actual Sales (₹)',    color='#666', fontsize=11, labelpad=10)
ax.set_ylabel('Predicted Sales (₹)', color='#666', fontsize=11, labelpad=10)
ax.set_title(f'Actual vs Predicted  —  R² = {r2:.4f}  |  RMSE = ₹{rmse:,.0f}',
             color='#f0ede6', fontsize=12, pad=14)

fmt = mticker.FuncFormatter(lambda x, _: f'₹{x/1000:.0f}k')
ax.xaxis.set_major_formatter(fmt)
ax.yaxis.set_major_formatter(fmt)
ax.tick_params(colors='#555', labelsize=9)
ax.grid(color='#1e1e1e', linewidth=0.6, zorder=0)
ax.legend(fontsize=9, framealpha=0, labelcolor='#888')

plt.tight_layout()
plt.savefig('actual_vs_predicted.png', dpi=150, bbox_inches='tight',
            facecolor='#0d0d0d')
plt.show()
print("Plot saved as actual_vs_predicted.png")


# ─────────────────────────────────────────────────────────────────────────────
# SECTION B  ·  Quantile Regression  (proper confidence intervals)
# Add directly after Section A cells.
# Trains two XGBRegressor models: Q10 (lower bound) and Q90 (upper bound).
# These replace the fixed ±MAE band in the Streamlit app.
# ─────────────────────────────────────────────────────────────────────────────

# Cell B1 — train quantile models
from xgboost import XGBRegressor

# Lower bound — 10th percentile
qr_low = XGBRegressor(
    objective='reg:quantileerror',
    quantile_alpha=0.10,
    n_estimators=300,
    learning_rate=0.05,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42
)
qr_low.fit(X_train, y_train)

# Upper bound — 90th percentile
qr_high = XGBRegressor(
    objective='reg:quantileerror',
    quantile_alpha=0.90,
    n_estimators=300,
    learning_rate=0.05,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42
)
qr_high.fit(X_train, y_train)

print("Quantile models trained ✓")


# Cell B2 — evaluate interval quality on test set
y_lower = qr_low.predict(X_test)
y_upper = qr_high.predict(X_test)

# Coverage: fraction of actuals that fall inside [lower, upper]
coverage = np.mean((y_test.values >= y_lower) & (y_test.values <= y_upper))

# Average interval width
avg_width = np.mean(y_upper - y_lower)

print(f"80% interval coverage : {coverage*100:.1f}%  (target ≥ 80%)")
print(f"Average interval width: ₹{avg_width:,.0f}")


# Cell B3 — visualise interval on a sample of 100 test points
sample_n   = 100
idx_sample = np.random.default_rng(0).choice(len(y_test), sample_n, replace=False)
idx_sorted = idx_sample[np.argsort(y_test.values[idx_sample])]  # sort by actual

act_s   = y_test.values[idx_sorted]
pred_s  = y_pred[idx_sorted]
low_s   = y_lower[idx_sorted]
high_s  = y_upper[idx_sorted]
x_range = np.arange(sample_n)

fig, ax = plt.subplots(figsize=(11, 4.5))
fig.patch.set_facecolor('#0d0d0d')
ax.set_facecolor('#131313')
for sp in ax.spines.values():
    sp.set_visible(False)

ax.fill_between(x_range, low_s, high_s,
                alpha=0.18, color='#4a7fcb', label='80% confidence interval')
ax.plot(x_range, act_s,  color='#f0ede6', linewidth=1.0,
        alpha=0.85, label='Actual sales', zorder=3)
ax.plot(x_range, pred_s, color='#ff6b35', linewidth=1.2,
        alpha=0.9,  label='Predicted (median)', zorder=4)

ax.set_xlabel('Test samples (sorted by actual sales)',
              color='#555', fontsize=10, labelpad=8)
ax.set_ylabel('Sales (₹)', color='#555', fontsize=10, labelpad=8)
ax.set_title('Quantile Regression — 80% Prediction Interval  '
             f'(coverage {coverage*100:.1f}%)',
             color='#f0ede6', fontsize=12, pad=14)

fmt = mticker.FuncFormatter(lambda x, _: f'₹{x/1000:.0f}k')
ax.yaxis.set_major_formatter(fmt)
ax.tick_params(colors='#555', labelsize=9)
ax.grid(color='#1e1e1e', linewidth=0.5, zorder=0)
ax.legend(fontsize=9, framealpha=0, labelcolor='#888')

plt.tight_layout()
plt.savefig('quantile_interval.png', dpi=150, bbox_inches='tight',
            facecolor='#0d0d0d')
plt.show()
print("Plot saved as quantile_interval.png")


# Cell B4 — save both quantile models (app.py will load them)
import joblib
joblib.dump(qr_low,  'bigmart_model_q10')
joblib.dump(qr_high, 'bigmart_model_q90')
print("bigmart_model_q10 saved ✓")
print("bigmart_model_q90 saved ✓")
```

```python
# ── Cell A: Install & import ──────────────────────────────────────────────────
# (shap is already in requirements.txt; this cell is just for local runs)
# !pip install shap --quiet

import shap
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# ── Cell B: Create & save the explainer ──────────────────────────────────────
#
# xg_final  → your trained XGBRFRegressor  (already defined above)
# final_data → the 5-column encoded DataFrame used for training
#
# TreeExplainer is the native, exact explainer for XGBoost — no approximation.

explainer = shap.TreeExplainer(xg_final)

# Compute SHAP values for the full training set (used for summary plots)
shap_values_train = explainer.shap_values(final_data)

# Save the explainer next to bigmart_model so app.py can load it
import joblib
joblib.dump(explainer, 'bigmart_explainer')
print("bigmart_explainer saved ✓")

# ── Cell C: Global summary plot (beeswarm) ────────────────────────────────────
# Shows how each feature affects predictions across all training samples.
# Red = high feature value pushes prediction up; blue = low value pulls it down.

plt.figure()
shap.summary_plot(
    shap_values_train,
    final_data,
    plot_type="dot",
    feature_names=["Item MRP", "Outlet ID", "Outlet Size", "Outlet Type", "Outlet Age"],
    show=False
)
plt.tight_layout()
plt.savefig("shap_summary_beeswarm.png", dpi=150, bbox_inches="tight")
plt.show()

# ── Cell D: Bar plot (mean absolute SHAP — feature importance ranking) ────────
plt.figure()
shap.summary_plot(
    shap_values_train,
    final_data,
    plot_type="bar",
    feature_names=["Item MRP", "Outlet ID", "Outlet Size", "Outlet Type", "Outlet Age"],
    show=False
)
plt.tight_layout()
plt.savefig("shap_summary_bar.png", dpi=150, bbox_inches="tight")
plt.show()

# ── Cell E: Dependence plot for Item_MRP (your #1 feature) ───────────────────
# Shows how Item_MRP's SHAP value changes across its value range,
# coloured by a second interacting feature (auto-selected by SHAP).

plt.figure()
shap.dependence_plot(
    "Item MRP",
    shap_values_train,
    final_data,
    feature_names=["Item MRP", "Outlet ID", "Outlet Size", "Outlet Type", "Outlet Age"],
    show=False
)
plt.tight_layout()
plt.savefig("shap_dependence_mrp.png", dpi=150, bbox_inches="tight")
plt.show()

# ── Cell F: Single-prediction waterfall (same logic as app.py) ───────────────
# Explain the first test-set prediction so you can verify the chart matches
# what the Streamlit app will show.

sample_idx    = 0
sample_input  = X_test.iloc[[sample_idx]]   # 1-row DataFrame
sample_sv     = explainer.shap_values(sample_input)[0]
feature_labels = ["Item MRP", "Outlet ID", "Outlet Size", "Outlet Type", "Outlet Age"]

order          = sample_sv.argsort()         # ascending absolute value

fig, ax = plt.subplots(figsize=(7, 3.2))
colors  = ['#ff6b35' if v > 0 else '#4a7fcb' for v in sample_sv[order]]
ax.barh([feature_labels[i] for i in order], sample_sv[order], color=colors, height=0.52)
ax.axvline(0, color='#aaaaaa', linewidth=0.8)
ax.set_xlabel("SHAP value (₹ impact on prediction)")
ax.set_title(
    f"Sample prediction: ₹{xg_final.predict(sample_input)[0]:,.0f}  "
    f"(actual: ₹{y_test.iloc[sample_idx]:,.0f})"
)
pos_p = mpatches.Patch(color='#ff6b35', label='Increases')
neg_p = mpatches.Patch(color='#4a7fcb', label='Decreases')
ax.legend(handles=[pos_p, neg_p], fontsize=8)
plt.tight_layout()
plt.savefig("shap_single_prediction.png", dpi=150, bbox_inches="tight")
plt.show()
```
